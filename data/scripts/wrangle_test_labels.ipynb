# Test Data Label Standardization

```python
import os
import sys
import pandas as pd

p = os.path.dirname(
    os.path.dirname(
    os.path.abspath(".")
    ))
print(p)
sys.path.append(p)
from src.server.fuzzy_match import DrugMatcher
```

```python
df_raw = pd.read_csv("test_labels.csv")
```

```python
fm = DrugMatcher("drug_dictionary_copy.csv", 0.8)
```

```python
df = df_raw.copy(deep=True)
s = pd.Series()
for idx, v in df.iterrows():
    r = fm.match(v["label"])
    if r["matched_label"]:
        s.loc[idx] = r["matched_label"]
    else:
        s.loc[idx] = ""
df["fm_label"] = s
```

```python
# unique labels sorted alphabetically
unique_labels = pd.Series(df["label"].str.lower().unique()).sort_values()
for d in unique_labels:
    print(d)
```

```python
# apply fuzzy match to unique labels
s = pd.Series()
for idx, v in unique_labels.items():
    r = fm.match(v)
    if r["matched_label"]:
        s.loc[idx] = r["matched_label"]
    else:
        s.loc[idx] = ""
dfu = pd.DataFrame({"label":unique_labels, "fm_label":s})
```

```python
# unique labels that do not match fuzzy matched labels
for idx, row in dfu.iterrows():
    if (row["label"].lower() != row["fm_label"].lower()):
        print(f"{row["label"]:20} {row["fm_label"]}", )
```

```python
# rows that are just injection
df1 = df[df["label"].str.lower().str.startswith("inj")]
print("count:", df1.shape[0])
print(df1[["filename", "label", "fm_label"]])
```

```python
# rows that refer to a tablet or are the word tablet
df2 = df[df["label"].str.lower().str.startswith("tab")]
print("count:", df2.shape[0])
print(df2[["filename", "label", "fm_label"]])
```

```python
# rows that refer to capsules or are just capsule
df3 = df[df["label"].str.lower().str.contains("cap")]
print("count:", df3.shape[0])
print(df3[["filename", "label"]])
```

```python
# rows labelled diet
df5 = df[df["label"].str.lower() == "diet"]
print("count:", df5.shape[0])
print(df5[["filename", "label", "fm_label"]])
```

```python
# rows that have dosage
df4 = df[df["has_dosage_flag"]]
# attempt to remove dosage
df4["clean_label"] = df4["label"].str.replace(r"\s+\d+", repl="", regex=True)
print("count:", df4.shape[0])
print(df4[["filename", "label", "fm_label", "clean_label"]])
```

```python
# rows with dosage that are not easy to remove
df7 = df[(df["has_dosage_flag"] & ~(df["label"].str.contains(r"\s+\d+$", regex=True)))]
print("count:", df7.shape[0])
for _, row in df7.iterrows():
    print(f"{row["filename"]:20} {row["label"]:15} {row["fm_label"]}")
```

## Apply Clean-Up

```python
df_clean = df.copy(deep=True)
```

```python
df_clean = df.copy(deep=True)

# remove rows where dosage is not formatted in a simple way
df_clean = df_clean[~(df_clean["has_dosage_flag"] & ~(df_clean["label"].str.contains(r"\s+\d+$", regex=True)))]
# remove dosage numbers where format is simple
df_clean["label"] = df_clean["label"].str.replace(r"\s+\d+", repl="", regex=True)
# remove rows where label is inj or injection
df_clean = df_clean[~(df_clean["label"].str.lower().str.startswith("inj"))]
# remove rows where label is tablet
df_clean = df_clean[~(df_clean["label"].str.lower() == "tablet")]
df_clean["label"] = df_clean["label"].str.lower().str.removeprefix("tab.").str.strip()
df_clean["label"] = df_clean["label"].str.lower().str.removeprefix("tb.").str.strip()
# remove rows where label is capsule
df_clean = df_clean[~(df_clean["label"].str.lower() == "capsule")]
df_clean["label"] = df_clean["label"].str.lower().str.removeprefix("cap.").str.strip()
# remove rows where label is diet
df_clean = df_clean[~(df_clean["label"].str.lower() == "diet")]
# remove rows where label is supp or suppository
df_clean = df_clean[~(df_clean["label"].str.lower().str.startswith("supp"))]
# remove rows where label is syrup
df_clean = df_clean[~(df_clean["label"].str.lower() == "syrup")]
print(df_clean.shape[0])
```

## Check labels that still do not match

```python
# add fuzzy match labels to df_clean
s = pd.Series()
for idx, v in df_clean.iterrows():
    r = fm.match(v["label"])
    if r["matched_label"]:
        s.loc[idx] = r["matched_label"]
    else:
        s.loc[idx] = ""
df_clean["fm_label"] = s
```

```python
dfm = df_clean[df_clean["label"].str.lower() != df_clean["fm_label"].str.lower()].sort_values("label")
print("count:", dfm.shape[0])
for _, row in dfm.iterrows():
    print(f"{row["filename"]:20} {row["label"]:15} {row["fm_label"]}")
```

## Manual Filtering

```python
# do not match and it is unclear what correct label should be or is a generic name
droplist = [ 
    "ceftazidime",
    "ceftriaxone",
    "cefuroxime",
    "cefuroximee",
    "devixil", 
    "dexter",
    "fexofenadine",
    "fexofenadinee",
    "high",
    "lactomore",
    "meropenem",
    "natumef",
    "nebulization",
    "normal",
    "ointment",
    "omeprazole",
    "vancomycin",
    "zymet",
    "seko",
    ]

# correct label is the one determined by the fuzzy match
keeplist = [
    "calboral", 
    "coralcal",
    "don a",
    "neuro",
    "neuro b",
    ]

# for each row in df_clean
# if not match
# manually select: keep or drop
df_final = df_clean.copy()
for idx, row in df_clean.iterrows():
    if row["label"].lower() != row["fm_label"].lower():
        print(f"{row["filename"]:20} {row["label"]:15} {row["fm_label"]}", end="")

        if row["label"] in droplist:
            df_final.drop(idx, inplace=True)
            print(" drop")
        elif row["label"] in keeplist:
            print(" keep")
        else:
            # manual
            k = input("1 keep, 2 drop")
            if k == "2":
                df_final.drop(idx, inplace=True)
                print(" drop")
            else:
                print(" keep")
```

```python
# display final dataframe
print("count:", df_final.shape[0])
for _, row in df_final.iterrows():
    print(f"{row["filename"]:20} {row["label"]:15} {row["fm_label"]}")
```

```python
# set standardized labels as label column for convenience
df_final["original_label"] = df_final["label"]
df_final["label"] = df_final["fm_label"]
df_final.to_csv("test_labels_standardized.csv")
```
